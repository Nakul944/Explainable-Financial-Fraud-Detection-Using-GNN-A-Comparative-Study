```python
import pandas as pd

features = pd.read_csv(
    "../datasets/Elliptic_Bitcoin/elliptic_txs_features.csv",
    header=None
)

classes = pd.read_csv(
    "../datasets/Elliptic_Bitcoin/elliptic_txs_classes.csv"
)

# first feature column = txId
# second column = time step
time_data = features[[0, 1]].copy()
time_data.columns = ["txId", "time_step"]

df = time_data.merge(
    classes,
    on="txId"
)

print(
    pd.crosstab(
        df["time_step"],
        df["class"]
    )
)
```
